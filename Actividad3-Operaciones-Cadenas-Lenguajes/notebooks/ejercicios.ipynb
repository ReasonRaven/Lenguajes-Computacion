# Actividad 3: Operaciones con cadenas y lenguajes

Este notebook resuelve los ejercicios usando los endpoints de la API FastAPI definida en `main.py`. Se usa `TestClient` para llamar a los endpoints reales sin necesidad de levantar un servidor aparte.

Datos del ejercicio: $\Sigma=\{a,b\}$, $L=\{\lambda,'a','b'\}$, $M=\{'b','aa'\}$. En la API $\lambda$ se representa como la cadena vacía `""`.

```python
import sys
sys.path.append("..")

from fastapi.testclient import TestClient
from main import app

client = TestClient(app)

def llamar(path, body):
    r = client.post(path, json=body)
    r.raise_for_status()
    return r.json()["resultado"]

L = ["", "a", "b"]
M = ["b", "aa"]
L, M
```

## Operaciones de conjuntos

### $L \cup M$

```python
L_union_M = llamar("/lenguajes/union", {"L": L, "M": M})
L_union_M
```

### $L \cap M$

```python
L_inter_M = llamar("/lenguajes/interseccion", {"L": L, "M": M})
L_inter_M
```

### $L - M$

```python
L_menos_M = llamar("/lenguajes/diferencia", {"L": L, "M": M})
L_menos_M
```

### $M - L$

```python
M_menos_L = llamar("/lenguajes/diferencia", {"L": M, "M": L})
M_menos_L
```

## Operaciones sobre cadenas (concatenación de lenguajes)

### $L.M$

```python
LM = llamar("/lenguajes/concatenacion", {"L": L, "M": M})
LM
```

### $M.L$

```python
ML = llamar("/lenguajes/concatenacion", {"L": M, "M": L})
ML
```

### $M^2 = M.M$

```python
M2 = llamar("/lenguajes/concatenacion", {"L": M, "M": M})
M2
```

## Clausura de Kleene y combinación

### $L^*$ (solo los primeros 8 elementos)

```python
L_star = llamar("/lenguajes/clausura_kleene", {"L": L, "max_elementos": 8})
L_star
```

### $M^*$ (solo los primeros 8 elementos)

```python
M_star = llamar("/lenguajes/clausura_kleene", {"L": M, "max_elementos": 8})
M_star
```

### $(LM) \cup (M^* \cap L^2)$

Primero se calcula $L^2 = L.L$, luego $M^* \cap L^2$, y finalmente su unión con $LM$.

```python
L2 = llamar("/lenguajes/concatenacion", {"L": L, "M": L})
L2
```

```python
M_star_inter_L2 = llamar("/lenguajes/interseccion", {"L": M_star, "M": L2})
M_star_inter_L2
```

```python
resultado_final = llamar("/lenguajes/union", {"L": LM, "M": M_star_inter_L2})
resultado_final
```
